```python
import pandas as pd
    
# 3. Read Files into Pandas
# Use engine='pyarrow' for faster/more memory-efficient Parquet reading
print("Reading data into Pandas...")
master_df = pl.scan_parquet("C:/Users/nitis/Desktop/Hotspot\Hotspotting PPI History - 20251020/hist_p.parquet")
updates_df = pl.read_csv("H:/Work/Hotspotting/Data/hotspotting_monthly_extract_new2.csv")
master_schema = master_df.collect_schema()  # Dictionary of {column: dtype}

# 2. Align columns and fix data types
# We iterate through the master schema and cast the CSV columns to match
transformation_expressions = []
for col_name, dtype in master_schema.items():
    if col_name in updates_df.columns:
        if dtype == pl.Date:
            # Force the string date into a real Date object
            expr = pl.col(col_name).str.to_date("%Y-%m-%d")
        else:
            # Force types (int64, float64, etc)
            expr = pl.col(col_name).cast(dtype)
        transformation_expressions.append(expr)

# Apply the transformations to the update stream
update_lazy_cleaned = updates_df.select(transformation_expressions)
```

```python
# Identify the earliest date in the update file
min_update_date = update_lazy_cleaned['end_date'].min()
max_update_date = "2018-01-01"

# Keep only the old data that is strictly BEFORE the updates start
master_df_f = master_df.filter(pl.col("end_date") < min_update_date)

master_df_f = master_df.filter(pl.col("end_date") > max_update_date)

updated_df = pl.concat([master_df_f, update_lazy_cleaned.lazy()]).collect()
```

```python
#Output the file
updated_df.write_parquet("C:/Users/nitis/Desktop/Hotspot\Hotspotting PPI History - 20251020/hist_u.parquet")
```
