```python
# Import the libraries
import pandas as pd
import numpy as np
from datetime import date

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
# Try reading the excel at some point
df_p = pd.read_csv("H:/Work/Hotspotting/Data/Hotspotting_monthly_metrics_202602.csv", engine='python')

#Data prepping step
df_p.fillna(0, inplace=True)

# Changing the date columns for easier joining later for Dwelling data
df_p['end_date'] = pd.to_datetime(df_p['end_date'])

# Sorting and filtering on the latest data for each property_type
df = df_p.sort_values(by=['suburb_code','property_type','end_date'])
```

```python
#Formats percentage
# List of all percentage columns to be processed
percentage_columns = [
    'sales_price_median_12mths_pctChange_3mths',
    'sales_price_median_12mths_pctChange_12mths',
    'sales_price_median_12mths_avgpctChange_5yrs',
    'asking_rent_12mths_pctChange_3mths',
    'asking_rent_12mths_pctChange_12mths',
    'Median Vendor Discount - 12 Months',
    'Median Estimated Rental Yield - Point in Time',
    'Sold below asking price - 12 Months',
    'Sold at asking price - 12 Months',
    'Sold above asking price - 12 Months',
    'Median Vendor Premium - 12 Months'
    ,'Vacancy Rate - 1 Month'
]

# Iterate through the columns and apply the simplified logic
pd.set_option('display.float_format', '{:.2f}'.format)
for col in percentage_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col] * 100
    df[col] = df[col].round(2)

#Manual steps for 'Vacancy Rate - 1 Month'
# df['Vacancy Rate - 1 Month'] = pd.to_numeric(df['Vacancy Rate - 1 Month'], errors='coerce')
# df['Vacancy Rate - 1 Month'] = df['Vacancy Rate - 1 Month'].round(2)
```

```python
# Modifying #'Currency' columns as requested:
# Note: All New column names here

# List of all columns provided
price_columns = [
    'Mean of For Sale Listing Prices - 12 Months', 'Median For Sale Listing Price - 12 Months',
    'Mean Sale Price - 12 Months', 'Median Sale Price - 1 Month', 'Median Sale Price - 12 Months',
    'Sale Price Percentile10 - 12 Months', 'Sale Price Percentile30 - 12 Months',
    'Sale Price Percentile70 - 12 Months', 'Sale Price Percentile90 - 12 Months',
    'Average advertised rental price - 12 Months', 'Median advertised rental price - 12 Months',
    'Advertised Rental Price Percentile10 - 12 Months', 'Advertised Rental Price Percentile25 - 12 Months',
    'Advertised Rental Price Percentile75 - 12 Months', 'Advertised Rental Price Percentile90 - 12 Months',
    'Median Estimated Rental Value - Point in Time', 'Estimated Rent Value Percentile10 - Point in Time',
    'Estimated Rent Value Percentile30 - Point in Time', 'Estimated Rent Value Percentile70 - Point in Time',
    'Estimated Rent Value Percentile90 - Point in Time', 'Estimated Sale Value Percentile20 - Point in Time',
    'Estimated Sale Value Percentile30 - Point in Time', 'Estimated Sale Value Percentile50 - Point in Time',
    'Estimated Sale Value Percentile70 - Point in Time', 'Estimated Sale Value Percentile90 - Point in Time',
    'Sale Price Percentile15 - 12 Months', 'Sale Price Percentile20 - 12 Months',
    'Sale Price Percentile25 - 12 Months', 'Sale Price Percentile35 - 12 Months',
    'Sale Price Percentile40 - 12 Months', 'Sale Price Percentile45 - 12 Months',
    'Sale Price Percentile50 - 12 Months', 'Sale Price Percentile55 - 12 Months',
    'Sale Price Percentile5 - 12 Months', 'Sale Price Percentile60 - 12 Months',
    'Sale Price Percentile65 - 12 Months', 'Sale Price Percentile75 - 12 Months',
    'Sale Price Percentile80 - 12 Months', 'Sale Price Percentile85 - 12 Months',
    'Sale Price Percentile95 - 12 Months','Total value of sales - 12 Months'
]

# Numeric Conversion

for col in price_columns:
    #Convert to Numeric
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Zero Decimals
for col in price_columns:
    # Round to the nearest whole number and convert to Pandas Nullable Integer type
    df[col] = df[col].round(0).astype(pd.Int64Dtype(), errors='ignore')

# Note following are unchanged:
# Suburb Name: As received
# Suburb Code: As received
# Postcode:  As received - confirmation by Johan
# LGA:  As received
# State  As received
# Property type: As received
# sales_volume_12mths: As received
# days_on_market_12mth: As received
# sale_demand_1mth: As received
# rental_demand_1mth: As received

#The following new columns are unchanged
# 'Days_on_Market_for_Sale_Percentile25___12_Months', 'Days_on_Market_for_Sale_Percentile75___12_Months', 'Mean_Hold_period___12_Months',
# 'bedrooms_group', 'Volume_of_Newly_Listed_Properties___12_Months', 'Volume_of_Listed_Properties___12_Months', 'Sold_above_asking_price___12_Months', 'Sold_at_asking_price___12_Months', 'Sold_below_asking_price___12_Months',
# 'Median_Days_on_Market_for_Leased_Properties___12_Months', 'Days_on_Market_for_Leased_Properties_Percentile10___12_Months', 'Days_on_Market_for_Leased_Properties_Percentile25___12_Months', 'Days_on_Market_for_Leased_Properties_Percentile75___12_Months', 'Days_on_Market_for_Leased_Properties_Percentile90___12_Months',
# 'Median_Days_on_Market_for_Sale___12_Months', 'Volume_of_rental_transactions___1_Month', 'Rental_Stock___1_Month', 'Volume_of_Vacant_Properties___1_Month', 'Volume_of_New_Properties_Listed_for_Rent___1_Month', 'Volume_of_Properties_Listed_for_Rent___1_Month',
# 'Volume_of_sales___12_Months', 'Median_Vendor_Premium___12_Months', 'Total_Demand_for_Properties_for_Sale___1_Month', 'Total_Demand_for_Properties_for_Rent___1_Month', 'Rent_Demand_to_Supply_Ratio___1_Month', 'Median_Demand_for_Properties_for_Rent___1_Month', 'Sale_Demand_to_Supply_Ratio___1_Month'
```

```python
# Dwelling vacancy rates and select the latest entry
# Removing DWELLING from the analysis and creating a dataset for DWELLING for Vacancy rate check later
df_dw = df[df['property_type'] == 'DWELLING'] #Create the vacancy rate field from Dwelling
df_dw = df_dw[['suburb_code','end_date','Vacancy Rate - 1 Month']] #'postcode'
#Rename vacancy in dwelling to identify
df_dw = df_dw.rename(columns={'Vacancy Rate - 1 Month': 'Vacancy Rate - 1 Month_dw'})

#Create the file for the rest of the data
df2 = df[df['property_type'] != 'DWELLING']

#Final file with the Dwelling column
df_fin = pd.merge(df2,df_dw, on=['suburb_code','end_date']) #,'postcode'"

# Sorting and filtering on the latest data for each property_type
df_fin2 = df_fin.sort_values(by=['suburb_code','property_type','end_date'])
df_fin3 = df_fin2.groupby(['suburb_code','property_type']).nth(-1).reset_index()
```

```python
# Output csv
df_fin3.to_csv("H:/Work/Hotspotting/Data/hotspotting_monthly_extract_new2.csv", #2 for Vacancy
    index=False,     # Do not include the DataFrame index as a column
    header=True,      # Include the column names (header)
)

# Excel output for Tim
df_fin3_ex = df_fin3.copy()

#Getting the percentage columns in the right decimals for the formatting in Excel
for col in percentage_columns:
    df_fin3_ex[col] = df_fin3_ex[col] / 100
    
df_fin3_ex.to_excel('H:/Work/Hotspotting/Data/latest_month_extract.xlsx', index=False)
```
